Vamos a evaluar si se requiere realizar limpieza de datos

```python
def fill_na(text_set):
    return text_set.fillna('').astype(str)
def normalize_text(text_set):
    return text_set.map(lambda x: unicodedata.normalize('NFKC', x)).str.lower()
def remove_spaces(text_set):
    return text_set.str.replace(r'\\s+', ' ', regex=True).str.strip()
def normalize_text_pipeline(text_set):
    return remove_spaces(normalize_text(fill_na(text_set)))

normalized_text = normalize_text_pipeline(raw_data['textos'])

quality_checks = pd.Series({
    'Textos faltantes': raw_data['textos'].isna().sum(),
    'Etiquetas faltantes': raw_data['ODS'].isna().sum(),
    'Textos vacíos': normalized_text.eq('').sum(),
    'Filas duplicadas': raw_data.duplicated().sum(),
    'Textos repetidos normalizados': normalized_text.duplicated().sum(),
    'Textos con etiquetas contradictorias': raw_data.assign(normalized_text=normalized_text).groupby('normalized_text')['ODS'].nunique().gt(1).sum(),
    'Etiquetas fuera de 1 a 17': raw_data['ODS'].between(1, 17).eq(False).sum()
})

display(raw_data.head())
display(quality_checks.to_frame('Cantidad'))
```

## BOW

Para la construcción de la bolsa de palabras, exploramos los distintos pasos necesarios para la preparación de texto el cual consta de tokenización, eliminación de stop-words y stemming.

Preparamos las instancias que vamos a necesitar

```python
ES_stopwords = set(stopwords.words('spanish'))
ES_stemmer = SnowballStemmer('spanish')
```

### Tokenización

Utilizamos un RegexpTokenizer, para hacer match de palabras utilizando expresiones regulares, eliminando así símbolos, espacios, números, y cualquier otro que no sea considerado una palabra.

```python
tokenizer = RegexpTokenizer(r'\w+')
tokenized = X_train.apply(lambda x: tokenizer.tokenize(x))
tokenized
```

### StopWords

Eliminamos las palabras vacías o stop-words utilizando el paquete en español de `stopwords`, con esto eliminamos aquellas palabras que no aporten un significado semántico a las frases como lo son los conectores o los artículos.

```python
no_stopwords = tokenized.apply(lambda x: [word for word in x if word.lower() not in ES_stopwords])
no_stopwords
```

### Nube de palabras

Llegados a este punto, podemos graficar una nube de palabras con la que podremos identificar las palabras con mayor peso en todo el corpus, se destacan las palabras "política", "embargo", "mujeres", "ejemplo", "sistema", "desarrollo", entre otras, dando a entender (al menos superficialmente) los temas principales del conjunto de datos.

```python
wc_text = ' '.join(word for text in no_stopwords for word in text)

wordcloud = WordCloud(
        width=800,
        height=400,
        background_color='white',
        min_font_size=10,
        max_font_size=110
    ).generate(wc_text)

plt.figure(figsize=(10, 10))
plt.imshow(wordcloud)
plt.axis('off')
plt.show()
```

### Stemming

Vamos a reducir cada token a su forma más general con el proceso de stemming, para este caso nuestro conjunto de datos está escrito en español, por lo que se opta por utilizar el `SnowballStemmer` pues tiene soporte para múltiples idiomas de acuerdo a la [documentaicón oficial](https://www.nltk.org/howto/stem.html).

```python
stemmed = no_stopwords.apply(lambda x: [ES_stemmer.stem(word) for word in x])
stemmed
```

### Texto procesado

Vamos a inspeccionar visualmente el resultado del proceso anterior, volviendo a armar las frases completas que ya fueron pre-procesadas. Esto nos ayuda a visualizar la forma en que serán interpretadas las frases por el modelo.

```python
processed = stemmed.apply(lambda x: ' '.join(x))
processed
```

Podemos condensar todo el proceso anterior en la siguiente función, la cual también nos será de utilidad para datos nuevos que se quieran procesar con el modelo entrenado.

```python
def text_analyzer(text):
    normalized = unicodedata.normalize('NFKC', text).lower()
    tokenized = tokenizer.tokenize(normalized)
    no_stopwords = [word for word in tokenized if word.lower() not in ES_stopwords]
    stemmed = [ES_stemmer.stem(word) for word in no_stopwords]
    
    return stemmed
```

### Nube de vocabularios

A continuación queremos graficar la nube de palabras con mayor peso generada por el TF-IDF Vectorizer, utilizando la función de pre procesamiento anterior

**Implementamos LSA**

Utilizando la clase de TruncatedSVD realizamos el agrupamiento de palabras con significados similares. Exploraremos la cantidad de componentes `n_components` [10, 15, 20], dentro del rango sugerido en el enunciado. Comparamos la varianza explicada acumulada. Esta exploración utiliza solo los textos de entrenamiento.

```python
cloud_vectorizer = TfidfVectorizer(analyzer=text_analyzer, max_features=500)
X_tfidf = cloud_vectorizer.fit_transform(X_train)

lsa_models = {}
lsa_summary = []
n_topics = [10, 15, 20]

for n in n_topics:
    lsa = TruncatedSVD(n_components=n, n_iter=10, random_state=STATE)
    lsa_model = lsa.fit(X_tfidf)
    lsa_models[n] = lsa_model
    lsa_summary.append({'Componentes': n, 'Varianza explicada': lsa_model.explained_variance_ratio_.sum()})

display(pd.DataFrame(lsa_summary))
lsa_20 = lsa_models[20]

def describe_topics(lsa_model, vocabulary, n_top_words=10):
    n_topics = 6
    rows = []
    
    for comp, weights in enumerate(lsa_model.components_[:n_topics], start=1):
        positive_weights = weights.argsort()[-n_top_words:][::-1]
        negative_weights = [index for index in weights.argsort()[:5] if weights[index] < 0]

        rows.append({
            'Componente': comp,
            'Palabras positivas': ', '.join(f'{vocabulary[i]} ({weights[i]:.2f})' for i in positive_weights),
            'Palabras negativas': ', '.join(f'{vocabulary[i]} ({weights[i]:.2f})' for i in negative_weights) or 'Ninguna'
        })
    
    return pd.DataFrame(rows)

topics = describe_topics(lsa_20, cloud_vectorizer.get_feature_names_out(), n_top_words=10)
display(topics)
```

Podemos evidenciar que a mayor número de componentes, mejor es la varianza, por lo que utilizar un TruncatedSVD con 20 componentes es lo mejor para este caso.

Se examinan los primeros seis componentes. Para cada uno se muestra las diez palabras con mayor peso positivo, sus coeficientes y las cinco palabras con peso más negativo, si existen.

Los coeficientes de SVD pueden tener ambos signos y no son probabilidades. Un componente puede expresar un contraste entre campos semánticos; cambiar el signo de todo el eje no altera la descomposición. Por eso se indican ambos polos y no se interpreta el peso negativo como irrelevancia de una palabra.

Ahora mostramos las palabras con mayor peso (loading) por componente, de los primeros 5 tópicos

```python
selected_topics = [0, 1, 2, 3, 4]

fig, axes = plt.subplots(1, 5, figsize=(22, 5))
terms = cloud_vectorizer.get_feature_names_out()

for ax, idx in zip(axes, selected_topics):
    weights = lsa_20.components_[idx]
    top_indices = weights.argsort()[::-1][:20]
    
    top_weights = {
        terms[i]: weights[i] for i in top_indices if weights[i] > 0
    }

    wc = WordCloud(
        width=400,
        height=400,
        background_color='white'
    ).generate_from_frequencies(top_weights)

    ax.imshow(wc, interpolation='bilinear')
    ax.set_title(f'Topic {ods_names[idx + 1]}', fontsize=16)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

Con esto podemos evidenciar que los temas de las ODS están alineados con el contenido de sus textos pre procesados. También podemos ver cómo hay temas que son compartidos entre distintos ODS, como lo es la palabra "mujer" o "trabajo".

## Modelo de clasificación

De acuerdo a un [ejercicio de clasificación de texto](https://scikit-learn.org/stable/auto_examples/text/plot_document_classification_20newsgroups.html) encontrado en scikit-learn, podemos observar que los benchmarks apuntan a que los modelos que mejor se desempeñaron son los siguientes:

- Desempeño en entrenamiento



- Desempeño en test



Como se puede apreciar, los mejores modelos para esto son LogisticRegression y ComplementNB (Naive Bayes). Para este caso utilizaremos LogisticRegression pues es el que mejor se acopla a los objetivos del proyecto, con el uso de TF-IDF y SVD.

Se usa regresión logística multiclase con regularización L2. Es una alternativa eficiente para vectores numéricos de baja dimensión, admite las coordenadas positivas y negativas de SVD y permite controlar el ajuste mediante C. Un valor pequeño de C implica mayor regularización.

El modelo aprende fronteras lineales en el espacio latente. Esto facilita su entrenamiento en CPU, aunque puede limitar la separación de ODS con vocabulario compartido. No se justifica comparar muchos algoritmos para cumplir el objetivo del proyecto.

```python
logreg = LogisticRegression(solver='lbfgs', max_iter=500, random_state=STATE)
```

### Creación del pipeline

Con las instancias definidas, definimos los pasos para el pipeline, con el vectorizador y el SVD. Definimos los parámetros de búsqueda que queremos explorar y definimos un modelo que utiliza SVD y otro que no lo utiliza, para validar el aporte de SVD a este ejercicio.

La normalización L2 de las coordenadas latentes y el clasificador controla la longitud de los vectores después de la proyección, que puede variar aunque el TF-IDF original estuviera normalizado.

```python
text_vectorizer = TfidfVectorizer(analyzer=text_analyzer, min_df=2, max_df=0.95, sublinear_tf=True)
text_svd = TruncatedSVD(n_iter=10, random_state=STATE)

steps = [
    ('vectorizer', text_vectorizer),
    ('svd', 'passthrough'),
    ('normalizer', Normalizer(norm='l2')),
    ('model', logreg),
]

C_values = [0.01, 0.1, 1, 10]
class_weight_values = [None, 'balanced']
n_components_values = [10, 15, 20]

param_grid = [
    {'svd': ['passthrough'], 'model__C': C_values, 'model__class_weight': class_weight_values},
    {'svd': [text_svd], 'svd__n_components': n_components_values, 'model__C': C_values, 'model__class_weight': class_weight_values}
]

kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=STATE)
```

Ahora creamos el objeto de pipeline con los pasos que se presentaron anteriormente, seleccionamos por **F1 macro**, que promedia el F1 de cada ODS con el mismo peso y evita que las clases grandes dominen la elección. También registramos accuracy como referencia. La opción **balanced** asigna pesos inversos a las frecuencias dentro del entrenamiento de cada ajuste; se compara con **None** porque no debe asumirse que ponderar siempre mejora el resultado.

```python
pipeline = Pipeline(steps)

search = GridSearchCV(
    pipeline,
    param_grid,
    scoring={'f1_macro': 'f1_macro', 'accuracy': 'accuracy'},
    refit='f1_macro',
    cv=kfold,
    n_jobs=-1
)
```

### Entrenamiento

Entrenamiento del modelo de clasificación utilizando el pipeline

```python
search.fit(X_train, y_train)
```

Verificamos los resultados del Grid Search ordenando por el rank del F1 macro mencionado anteriormente

```python
cv_results = pd.DataFrame(search.cv_results_)
display(cv_results.sort_values(by='rank_test_f1_macro', ascending=True).head(10))
```

Como podemos observar, el mejor modelo no requiere de utilizar SVD

```python
best_model = search.best_estimator_
print('Best model parameters:', search.best_params_)
```

Los mejores parámetros son C=10, sin pesos inversos y omitiendo totalmente el SVD, por lo que podemos concluir que el uso de SVD para este caso no beneficia al modelo.

## Evaluación del modelo

Vamos a evaluar el modelo con los datos de test

```python
y_pred = best_model.predict(X_test)
```

Vamos a ver la matriz de confusion

```python
labels = best_model.classes_
display_labels = [f'{idx}. {ods_names[idx]}' for idx in labels]
cm = confusion_matrix(y_test, y_pred, labels=labels)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=display_labels)
disp.plot(xticks_rotation=90)
plt.title('Matriz de Confusión', fontsize=16)
plt.show()
```
