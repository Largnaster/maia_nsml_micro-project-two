## BOW

Creamos la bolsa de palabras con el siguiente pipeline que consta de tokenización, eliminación de stop-words y stemming.

### Tokenización

```python
tokenizer = RegexpTokenizer(r'\w+')
tokenized = X_train.apply(lambda x: tokenizer.tokenize(x))
tokenized
```

### StopWords

Eliminamos las palabras vacías o stop-words

```python
es_stopwords = stopwords.words('spanish')
no_stopwords = tokenized.apply(lambda x: [word for word in x if word.lower() not in es_stopwords])
no_stopwords
```

### Nube de palabras

```python
wc_text = ' '.join(word for text in no_stopwords for word in text)

wordcloud = WordCloud(
        width=800,
        height=400,
        background_color='white',
        min_font_size=10,
        max_font_size=110
    ).generate(wc_text)

plt.figure(figsize=(10, 10))
plt.imshow(wordcloud)
plt.axis('off')
plt.show()
```

### Stemming

Vamos a reducir al token a su forma más general con el proceso de stemming, para este caso, con un corpus en español, se opta por utilizar el `SnowballStemmer` pues tiene soporte para múltiples idiomas de acuerdo a la [documentaicón oficial](https://www.nltk.org/howto/stem.html).

```python
stemmer = SnowballStemmer('spanish')
stemmed = no_stopwords.apply(lambda x: [stemmer.stem(word) for word in x])
stemmed
```

### Texto procesado

Volvemos a reconstruír las cadenas de texto para que pueda ser procesada

```python
processed = stemmed.apply(lambda x: ' '.join(x))
processed
```

La función que realiza todo el procesamiento se encuentra a continuación

```python
def text_preprocessing(text):
    tokenizer = RegexpTokenizer(r'\w+')
    tokenized = tokenizer.tokenize(text)
    
    es_stopwords = stopwords.words('spanish')
    no_stopwords = [word for word in tokenized if word.lower() not in es_stopwords]
    
    stemmer = SnowballStemmer('spanish')
    stemmed = [stemmer.stem(word) for word in no_stopwords]
    
    processed_text = ' '.join(stemmed)
    
    return processed_text
```

Vectorizer, lo creamos a partir de la función de pre procesamiento anterior

```python
vectorizer = TfidfVectorizer(preprocessor=text_preprocessing)
X_tfidf = vectorizer.fit_transform(X_train)
X_tfidf
```

### Implementamos LSA

Utilizando la clase de TruncatedSVD realizamos el agrupamiento de palabras con significados similares. Seguiremos lo que nos sugiere la documentación y utilizaremos el número recomendado de componentes para análisis de textos (`n_components=100`)

```python
n_topics = 15

lsa = TruncatedSVD(n_components=n_topics, algorithm='randomized', random_state=STATE)
X_lsa = lsa.fit_transform(X_tfidf)

print(f'Varianza explicada por cada componente: {lsa.explained_variance_ratio_.sum():.2%}')
```

Mostramos las palabras con mayor peso (loading) por componente

```python
selected_topics = [0, 1, 2, 3, 4]

fig, axes = plt.subplots(1, 5, figsize=(22, 5))
terms = vectorizer.get_feature_names_out()

for ax, idx in zip(axes, selected_topics):
    topic = lsa.components_[idx]
    top_indices = np.argsort(np.abs(topic))[::-1][:20]
    
    top_weights = {terms[i]: abs(topic[i]) for i in top_indices}

    wc = WordCloud(
        width=400,
        height=400,
        background_color='white'
    ).generate_from_frequencies(top_weights)

    ax.imshow(wc, interpolation='bilinear')
    ax.set_title(f'Topic {idx + 1}', fontsize=16)
    ax.axis('off')

plt.tight_layout()
plt.show()
```

## Modelo de clasificación

De acuerdo a un [ejercicio de clasificación de texto](https://scikit-learn.org/stable/auto_examples/text/plot_document_classification_20newsgroups.html) encontrado en scikit-learn, podemos observar que los benchmarks apuntan a que los modelos que mejor se desempeñaron son los siguientes:

- Desempeño en entrenamiento



- Desempeño en test



Como se puede apreciar, los mejores modelos para esto son LogisticRegression y ComplementNB (Naive Bayes)

```python
nb = ComplementNB(alpha=0.1)
logreg = LogisticRegression(random_state=STATE, max_iter=500)
```

### Creación del pipeline

Con las instancias definidas, definimos los pasos para el pipeline

```python
steps = [
    ('vectorizer', vectorizer),
    ('model', logreg),
]
```

Ahora definimos el objeto de pipeline con los pasos

```python
pipeline = Pipeline(steps)
```

### Entrenamiento

Entrenamiento del modelo de clasificación utilizando el pipeline

```python
pipeline.fit(X_train, y_train)
```

## Evaluación del modelo

Vamos a evaluar el modelo con los datos de test

```python
y_pred = pipeline.predict(X_test)
```

Vamos a ver la matriz de confusion

```python
cm = confusion_matrix(y_test, y_pred)
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=pipeline.classes_)
disp.plot(xticks_rotation=45)
plt.title('Matriz de Confusión', fontsize=16)
plt.show()
```

Y ahora veremos el reporte de clasificación

```python
print(classification_report(y_test, y_pred))
```
